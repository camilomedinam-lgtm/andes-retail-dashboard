#### 3.1.1. Vista Overview — ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?

**KPIs incluidos y justificación** (5 tarjetas individuales, ancho uniforme, fuente 28-32pt y títulos ocultos):
- **Ingresos:** Muestra el volumen total de dinero generado.
- **Margen ($):** Complementa el margen porcentual con la magnitud absoluta de rentabilidad generada — da contexto financiero directo junto al ingreso total.
- **Margen %:** Asegura la visibilidad de la rentabilidad relativa del negocio.
- **Ticket Promedio:** Indica el valor medio de cada transacción.
- **Unidades Vendidas:** Permite entender el volumen físico movido.

**Gráficos elegidos y justificación:**
- **Gráfico de tiempo (Líneas / Dual Axis):** Mes vs Ingresos y Margen. Permite identificar tendencias y estacionalidad en ventas, cruzadas directamente con la salud del margen.
- **Gráfico comparativo geográfico, de categoría y de segmento (Barras apiladas):** Ingresos por País, Categoría de Producto y Segmento Cliente, con el Margen resaltado dentro de cada barra (naranja) sobre el Costo (gris). Para identificar de un vistazo no solo qué mercado, categoría o segmento factura más, sino cuál es proporcionalmente más rentable.

**Filtros:**
- Fecha Pedido
- País
- Región
- Categoría Producto
- Segmento Cliente


#### 3.1.2. Vista Detalle — ¿Por qué se han caído/subido/mantenido los ingresos?

**KPIs incluidos y justificación** (5 tarjetas, ancho uniforme):
- Se conservan Ingresos, Margen ($), Margen % y Ticket Promedio para mantener el mismo contexto financiero que en Overview y permitir comparación directa entre vistas.
- Se reemplaza la tarjeta de Unidades Vendidas por el % de Venta Alta. Justificación: al profundizar en el análisis, medir el volumen físico pierde relevancia; este nuevo KPI permite evaluar la calidad de nuestras ventas y la eficiencia comercial de los tickets.

**Gráficos elegidos y justificación:**
- **Heatmap (Mapa de calor):** Categoría Producto vs Estación (Color por Ingresos). Ideal para detectar rápidamente qué productos se venden mejor en qué época del año.

### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 5.1.1. 🖥️ Vista General (Overview)

**S (Situación):**
Andes Retail Group generó $5.53M en ingresos totales durante 2024–2025, operando en tres mercados (Colombia, Perú y Chile) con un margen bruto saludable de 35.1%. El negocio muestra un patrón estacional muy marcado y consistente: los ingresos alcanzan su pico en diciembre-enero y su valle en junio-julio, repitiéndose casi idéntico en ambos años.

**C (Complicación):**
Pese a mantener el mismo ritmo estacional, el ingreso total cayó de 2.86 M en 2024 a 2.67M en 2025, una retracción de 6.8% interanual. La caída no está localizada: los tres países retrocedieron (Perú, el mercado líder, es el más afectado con -10.3%; Colombia -4.5%; Chile -4.4%), y tres de las cuatro categorías de producto también cayeron, con Hogar como única excepción (+3.3%).

**Q (Pregunta):**
Si la estacionalidad se mantiene estable, ¿qué está impulsando esta caída transversal en ingresos? ¿Es un problema de menos pedidos, de tickets más pequeños, o de ambos? ¿Hay alguna palanca comercial que compense la tendencia?

**A (Respuesta):**
La Vista Detalle muestra que el problema no es de precio sino de volumen de actividad comercial de alto valor — el foco debe estar en aumentar la proporción de pedidos "Venta Alta", no en bajar precios ni en perseguir más unidades sueltas.

---

#### 🔎 Vista Detalle

**S (Situación):**
Al clasificar cada pedido con Nivel_Venta (Alta si Ingresos ≥ 1,000), el 44% de los pedidos ("Venta Alta") concentra el 72% del ingreso total (3.98M de 5.53M). Estos pedidos tienen en promedio un precio unitario 59% más alto (121 vs 77) y casi el doble de unidades por pedido (15.6 vs 8.3) que los de "Venta Baja".

**C (Complicación):**
La correlación entre precio unitario y unidades vendidas es prácticamente nula (-0.01), no hay evidencia de que el cliente compre menos cuando el precio sube. Esto descarta que la caída interanual se explique por sensibilidad al precio: no es que estemos vendiendo más caro y por eso menos, simplemente hay menos actividad comercial de alto valor.

**Q (Pregunta):**
Si no es un problema de precio, ¿dónde se concentra la pérdida de actividad. Es un tema de retención de clientes recurrentes, de adquisición de clientes nuevos, o de concentración de riesgo en pocas cuentas?

**A (Respuesta):**
La base de clientes está sana y muy distribuida: de 3,821 clientes únicos, los 5 principales representan apenas el 0.8% del ingreso total, así que no hay dependencia de cuentas clave ni riesgo de concentración. El foco comercial del próximo trimestre debe ser táctico y transversal: campañas de cross-sell/bundle que empujen pedidos de "Venta Baja" hacia el umbral de $1,000, priorizando Perú por su peso en el ingreso total y Sur por ser la región con la caída más pronunciada.

**📊 Actualización desempeño comercial**

Equipo, cerramos 2024–2025 con 5.53M en ingresos y un margen saludable de 35.1%, pero hay una señal de alerta: **los ingresos cayeron 6.8% interanual** (de 2.86M en 2024 a 2.67M en 2025), y la caída es transversal ya que los 3 países retroceden (Perú, nuestro mercado líder, es el más golpeado con -10.3%) y 3 de 4 categorías también, con Hogar como única excepción (+3.3%).
